```python


import requests

url = 'http://openapi.seoul.go.kr:8088/46487a6675746f703539496e596944/json/subwayStationMaster/1/100/'

response = requests.get(url)
print(response.content)

```

```python
data = response.json()
data
```

```python
import pandas as pd
temp = []

for d in data ['subwayStationMaster']['row']:
  temp.append(d)
a = pd.DataFrame(temp)
a
```

```python
import folium
```

```python

b = folium.Map(location=[37.448028, 127.169366])
b
```

```python
from folium.plugins import MarkerCluster
mc = MarkerCluster().add_to(b)

for _, row in a.iterrows():
    folium.Marker(location=[row["LAT"], row["LOT"]],
                  popup=folium.Popup(row["BLDN_NM"], max_width=200)
                 ).add_to(mc)

b
```
